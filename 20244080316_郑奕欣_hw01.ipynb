```python
# ===== 第二题：欧几里得算法 =====

# 1. 实现 gcd 函数
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 2. 评估指定数字对的 GCD
print("(2018, 2019) 的 GCD:", gcd(2018, 2019))
print("(1200, 300) 的 GCD:", gcd(1200, 300))
print("(5040, 60) 的 GCD:", gcd(5040, 60))

# 3. 测试负数参数的函数表现
print("(-2018, 2019) 的 GCD:", gcd(-2018, 2019))
print("(-1200, -300) 的 GCD:", gcd(-1200, -300))

# 【第3小问回答】：
# 当参数包含负数时，该函数仍然会返回一个非负的最大公约数（GCD）。
# 例如 gcd(-2018, 2019) 返回 1，gcd(-1200, -300) 返回 300。
# 
# 这种行为是有意义的。因为在数学定义中，最大公约数通常指正整数。
# Python 中的取模运算（%）结果符号与除数一致，这使得欧几里得算法在遇到负数时，
# 依然能够正确收敛并求出绝对值的最大公约数，即 gcd(m, n) = gcd(|m|, |n|)。
```

```python
import math

# ===== 第三题：近似欧拉数 e =====

# 1. euler_limit
def euler_limit(n):
    return (1 + 1 / n) ** n

# 2. euler_infinite_sum
def euler_infinite_sum(n):
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1 / factorial
    return total

# 3. euler_approx
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    factorial = 1
    k = 0
    while abs(total - true_e) > epsilon:
        if k > 0:
            factorial *= k
        total += 1 / factorial
        k += 1
    return total

# 4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

# ===== 测试部分 =====
print("--- 测试 euler_limit ---")
print(f"euler_limit(1000) = {euler_limit(1000)}")

print("\n--- 测试 euler_infinite_sum ---")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}") # 应该是 2.666...

print("\n--- 测试 euler_approx ---")
print(f"euler_approx(1e-6) = {euler_approx(1e-6)}")

print("\n--- 测试 print_euler_sum_table ---")
print_euler_sum_table(5)
```

```python
# ===== 第四题：测试整数属性 =====

# 1. 判断是否是 2 的幂
def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

# 2. 推广到判断是否是 b 的幂
def is_power(b, n):
    if b < 1 or n < 1:
        return False
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1

# ===== 测试部分 =====
print("--- 测试 is_power_of_2 ---")
print(f"is_power_of_2(1): {is_power_of_2(1)}")     # True (2^0)
print(f"is_power_of_2(8): {is_power_of_2(8)}")     # True (2^3)
print(f"is_power_of_2(12): {is_power_of_2(12)}")   # False
print(f"is_power_of_2(1024): {is_power_of_2(1024)}") # True (2^10)

print("\n--- 测试 is_power ---")
print(f"is_power(3, 27): {is_power(3, 27)}")       # True (3^3)
print(f"is_power(2, 16): {is_power(2, 16)}")       # True (2^4)
print(f"is_power(2, 18): {is_power(2, 18)}")       # False
print(f"is_power(1, 1): {is_power(1, 1)}")         # True (1^0 = 1)
print(f"is_power(1, 5): {is_power(1, 5)}")         # False
```
